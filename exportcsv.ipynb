```python
import pandas as pd
from sqlalchemy import create_engine

users = pd.read_csv("cleaned_data/users_cleaned.csv")
movies = pd.read_csv("cleaned_data/movies_cleaned.csv")
watch = pd.read_csv("cleaned_data/watch_cleaned.csv")
recommendations = pd.read_csv("cleaned_data/recommendations_cleaned.csv")
search = pd.read_csv("cleaned_data/search_cleaned.csv")
reviews = pd.read_csv("cleaned_data/reviews_cleaned.csv")

# Format: postgresql://username:password@host:port/database_name
engine = create_engine(
    "postgresql://postgres:postgres1@localhost:5432/netflix_churn"
)

# users.to_sql(
#     name="Users",
#     con=engine,
#     if_exists="replace",
#     index=False,
# )

movies.to_sql(
    name="Movies",
    con=engine,
    if_exists="replace",
    index=False,
)

watch.to_sql(
    name="Watch History",
    con=engine,
    if_exists="replace",
    index=False,
)

recommendations.to_sql(
    name="Recommendations",
    con=engine,
    if_exists="replace",
    index=False,
)

search.to_sql(
    name="Search History",
    con=engine,
    if_exists="replace",
    index=False,
)

reviews.to_sql(
    name="Reviews History",
    con=engine,
    if_exists="replace",
    index=False,
)

print("Data uploaded successfully!")
```
